```python
df = pd.read_excel("base de datos programación proyecto.xlsx", header=1)
```

```python
df = df.dropna()
```

```python
df.shape
```

```python
ventas_genero=df.groupby("Genre")["Global_Sales"].sum()

ventas_genero.sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

ventas_genero.sort_values().plot(kind="barh")

plt.title("Ventas globales por género")
plt.xlabel("Ventas globales")
plt.ylabel("Género")
plt.show()
```

```python
ventas_plataforma=df.groupby("Platform")["Global_Sales"].sum()

ventas_plataforma.sort_values(ascending=False)
```
